```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregar dados
df = pd.read_csv('dataset_consumo.csv')

# Visualizar dados
print(df.head())

# Converter mês
df['mes'] = pd.to_datetime(df['mes'])

# Ordenar
df = df.sort_values('mes')

# Gráfico de consumo
plt.figure()
plt.plot(df['mes'], df['consumo_kwh'])
plt.title('Consumo de Energia ao Longo do Tempo')
plt.xticks(rotation=45)
plt.show()

# Gráfico de valor da fatura
plt.figure()
plt.plot(df['mes'], df['valor_fatura'])
plt.title('Valor da Fatura ao Longo do Tempo')
plt.xticks(rotation=45)
plt.show()

# Insights simples
print("Consumo médio:", df['consumo_kwh'].mean())
print("Maior consumo:", df['consumo_kwh'].max())
```
